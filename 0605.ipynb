```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')

from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
import unidecode
```

# Limpeza de Dados Textuais

```python
data = [
    "O rato roeu a roupa do rei de Roma.",
    "A raposa marrom rápida pula sobre o cachorro preguiçoso. A raposa é rápida e inteligente.",
]

# Tokenization level
print("Tokenized Data:")
print([sent_tokenize(d) for d in data])

print("Word Tokenization")
print([word_tokenize(d) for d in data])
```

```python
print("Stop Words")
stop_words = set(stopwords.words('portuguese'))
# print(stop_words)

tokenized_data = [word_tokenize(d) for d in data]
tokenized_data_no_stp = [[word for word in d if word.lower() not in stop_words] 
                         for d in tokenized_data]
print ("Tokenized data com stop words:")
print(tokenized_data)
print ("Tokenized data sem stop words:")
print(tokenized_data_no_stp)
```

# Normalização de Texto

```python
# Conversão de Caixa

print("Transformação em caixa baixa (minúsculo)")
tokenized_data_lower = [[word.lower() for word in d] for d in tokenized_data_no_stp]
print(tokenized_data_lower)

print("Transformação em caixa alta (maiúsculo)")
tokenized_data_upper = [[word.upper() for word in d] for d in tokenized_data_no_stp]
print(tokenized_data_upper)
```

# Stemmatização

```python
# Stemming

stemmer = PorterStemmer()
semmed_data = [[stemmer.stem(word) for word in d] for d in tokenized_data]
print("Stemmed data:")
print(semmed_data)
```

# Lematização

```python
lemma = nltk.WordNetLemmatizer()
lemmatized_data = [[lemma.lemmatize(word) for word in d] for d in tokenized_data]
print("Lemmatized data:")
print(lemmatized_data)
```

# Removendo acentos

```python
text = "Olá, como você está? Eu estou bem, obrigado!"

print("Remove acentos:")
no_accents = [unidecode.unidecode(word) for word in text.split()]
print(no_accents)
```

# Tratando Pontuações 
- Perceba que o tokenizer já trata as pontuações, deixando elas isoladas das palavras.

```python
import string

print(string.punctuation)

text = 'Hello, how are you? I am fine, thank you!'

print("Remove pontuação:")
no_punctuation = ''.join([c for c in text if c not in string.punctuation])
print(no_punctuation)
```

# Remoção de caracteres repetidos

```python
text = "This is a    texxt withg repeeeeeeeted chaarrrraaaacters expreeeesion"

no_extra_spaces = ' '.join(text.split())
print("Remove espaços extras:")
print(no_extra_spaces)

import re
no_repeated_characters = re.sub(r'(.)\1+', r'\1', text)
print("Remove caracteres repetidos:")
print(no_repeated_characters)
```

# Expansão de abreviações e Termos

```python
texto = "vc é um cara legal que trabalha com ML e IA"

abreviacoes = {
    "vc": "você",
    "ML": "machine learning",
    "IA": "inteligência artificial"
}

texto_expandido = ' '.join([abreviacoes[word] 
                            if word in abreviacoes 
                            else word for word in texto.split()])
print("Texto expandido:")
print(texto_expandido)
```

# Utilização de Expressões Regulares

```python
import re
import string

re_url = re.compile(r'https?://\S+')
re_email = re.compile(r'\S+@\S+')


text = 'This is a text with a URL https://www.example.com and an email user@example.com'

print("Remove URLs:")
print(re_url.sub('', text))

print("Remove emails:")
print(re_email.sub('', text))

print("Removing Punctuation:")
print(re.sub(f'[{re.escape(string.punctuation)}]', '', text))
```
