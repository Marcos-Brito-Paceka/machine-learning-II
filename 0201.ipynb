# Ativação da Camada 1

$a^{1} = 1/(1+e^{-z})$

```python
import numpy as np
```

```python
class MLP():
    def __init__(self):
        self.theta1 = np.array([[0.5, 0.2],
                                [0.6, -0.1],
                                [-0.4, -0.3]])
        self.theta2 = np.array([[0.7, -0.1, 0.2]])

    def theta1(self):
        return self.theta1

    def theta2(self):
        return self.theta2

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def z1(self, x):
        return np.matmul(x, self.theta1.T) 
    
    def forward(self, x):
        z1 = np.matmul(x, self.theta1.T)
        a1 = self.sigmoid(z1)
        z2 = np.matmul(a1, self.theta2.T)
        a2 = self.sigmoid(z2)
        return a2
    
    def loss(self, y_pred, y_true):
        return np.mean((y_pred - y_true)**2)
    

model = MLP()
```

```python
X = np.array([[0.5, 0.1], 
              [0.2, 0.6]])

model.forward(X)
```

## Backward Pass

```python
Y = np.array([[0.7], 
              [0.8]])
a2 = model.forward(X)

delta2 = (a2 - Y) * (a2 * (1 - a2))

delta2
```

```python
z1 = model.z1(X)
print("z1:", z1)
a1 = model.sigmoid(z1)
print("a1:", a1)

delta1 = np.matmul(delta2, model.theta2) * (a1 * (1 - a1))

delta1
```

```python
grad_theta2 = np.matmul(delta2.T, a1)
print("grad_theta2:", grad_theta2)

grad_theta1 = np.matmul(delta1.T, X)
print("grad_theta1:", grad_theta1)
```

## Optmizer

$\theta_{t+1} = \theta_{t} - \eta \cdot \nabla J$

```python
theta1 = model.theta1 - 0.1 * grad_theta1
theta2 = model.theta2 - 0.1 * grad_theta2

theta1

```

```python
# calculando novamente a partir dos novos pesos

model.theta1 = theta1
model.theta2 = theta2
```

```python
model.forward(X)
```

```python
J = model.loss(model.forward(X), Y)
J
```
