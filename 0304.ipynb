```python
from torchvision.datasets import MNIST
from torchvision.transforms import ToTensor    
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader
```

```python

dataset_train = MNIST(root='data/',train=True, download=True, transform=ToTensor())
dataset_test = MNIST(root='data/',train=False, download=True, transform=ToTensor())
```

```python
uma_imagem = dataset_train.data[5, :, :]
plt.imshow(uma_imagem , cmap='gray')
```
