```python
from ultralytics import YOLO
```

```python
# carrega um modelo oficial
model = YOLO("yolov8x.pt") 
```

```python
# Realiza predições em uma imagem
results = model("https://ultralytics.com/images/bus.jpg") 
results[0].save('./results/bus_detected.jpg')
```

```python
# Itera sobre os resultados para mostrar as bounding boxes
for result in results:
    # bounding boxes no formato [xmin, ymin, xmax, ymax]
    boxes = result.boxes.xyxy.cpu().numpy()  

    # Labels das classes
    labels = [model.names[int(cls)] for cls in result.boxes.cls]

    # Obtem as confianças das predições
    confidences = result.boxes.conf.cpu().numpy()

    # Faz print das predições
    for box, label, confidence in zip(boxes, labels, confidences):
        print(f"Box: {box}, Label: {label}, Confidence: {confidence}")
```
