# Criando um Classificador com TF-IDF

## Importando as bibliotecas necessárias

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer, TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import cross_val_score
import pandas as pd
```

## Aplicando o TF-IDF
- Existem duas classes no Scikit-Learn para aplicar o TF-IDF: TfidfTransformer e TfidfVectorizer
- TfidfTransformer: Transforma uma matriz de contagem de termos em uma matriz de valores TF-IDF
- TfidfVectorizer: Combina as funções de CountVectorizer e TfidfTransformer em uma única classe

```python
documents = [
    "This is a simple example of Bag of Words.",
    "In natural language processing, Bag of Words is a basic technique.",
    "Bag of Words represents text data in the form of word occurrence counts.",
    "This is is a complex example of Bag of Words.",
]
```

```python
bow = CountVectorizer()
vec_bow = bow.fit_transform(documents)
pd.DataFrame(vec_bow.toarray(), columns=bow.get_feature_names_out())
```

```python
tfidf = TfidfTransformer()
vec_tfidf = tfidf.fit_transform(vec_bow)
pd.DataFrame(vec_tfidf.toarray(), columns=bow.get_feature_names_out())
```

```python
tfidf_vec = TfidfVectorizer()
vec_tfidf_2 = tfidf_vec.fit_transform(documents)
pd.DataFrame(vec_tfidf_2.toarray(), columns=tfidf_vec.get_feature_names_out())
```

## Carregando o dataset 20 News Groups

```python
train_data = fetch_20newsgroups(subset='train',data_home='./data/train', remove=('headers', 'footers', 'quotes'))
test_data = fetch_20newsgroups(subset='test',data_home='./data/test', remove=('headers', 'footers', 'quotes'))
```

```python
train_df = pd.DataFrame.from_dict({'text':train_data.data,'target':train_data.target})
test_df = pd.DataFrame.from_dict({'text':test_data.data,'target':test_data.target})
```

## Aplicando Pré-processamento nos dados

```python
import re
import string
re_url = re.compile(r'https?://\S+')
re_email = re.compile(r'\S+@\S+\.\S+')

def clean_text(text):
    text = text.lower().strip()
    text = re.sub(re_url, 'URL', text)
    text = re.sub(re_email, 'EMAIL', text)
    text = re.sub(f'[{re.escape(string.punctuation)}]', '', text)
    # text = re.sub(r'\d+', ' ', text)
    # text = re.sub(r'\s+', ' ', text)
    return text
```

```python
train_df.head()
```

```python
train_df['text'] = train_df['text'].apply(clean_text)
test_df['text'] = test_df['text'].apply(clean_text)
```

## Aplicando o CountVectorizer

```python
cv = CountVectorizer()

X_train = cv.fit_transform(train_df['text'])

X_train
```

```python
print(cv.get_feature_names_out())
```

```python
cv = CountVectorizer(min_df=0.001, lowercase=True, stop_words='english')

X_train = cv.fit_transform(train_df['text'])

X_train
```

## Construindo o pipeline

```python
bow = CountVectorizer(min_df=0.001, lowercase=True, stop_words='english')
tfidf = TfidfTransformer()
model = LogisticRegression()

bow_pipe = Pipeline([('bow', bow),('model', model)])

tfidf_pipe = Pipeline([('bow', bow), ('tfidf', tfidf),('model', model)])
```

## Treinando modelo com o pipeline

```python
resultados_bow = cross_val_score(bow_pipe, train_df['text'], train_df['target'], cv=5, n_jobs=-1, scoring='accuracy')
print(resultados_bow.mean())
```

```python
resultados_tfidf = cross_val_score(tfidf_pipe, train_df['text'], train_df['target'], cv=5, n_jobs=-1, scoring='accuracy')
print(resultados_tfidf.mean())
```
