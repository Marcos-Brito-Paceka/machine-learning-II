```python
import numpy as np
import pandas as pd
```

## Criando os Documentos

```python
documents = [
    "This is a simple example of Bag of Woerds.",
    "In natural language processing, Bag of Words is a basic technique.",
    "Bag of Words represents text data in the form of word occurrence counts.",
    "This is is a complex example of Bag of Words.",
]
```

## Criando a Classe BagOfWords

```python
class BagOfWords:
    def __init__(self):
        self.vocab = {}
        self.bow = []

    def fit_transform(self, documents):
        self.__build_vocab(documents)
        self.__create_bow_matrix(documents)
        return self.bow

    def __build_vocab(self, documents):
        for d in documents:
            for word in d.split():
                if word not in self.vocab:
                    self.vocab[word] = len(self.vocab)

    def __create_bow_matrix(self, documents):
        for d in documents:
            d_vec = [0] * len(self.vocab)
            for word in d.split():
                d_vec[self.vocab[word]] += 1
            self.bow.append(d_vec)

    def get_feature_names(self):
        return list(self.vocab.keys())
```

```python
custom_bow = BagOfWords()

bow_matrix = custom_bow.fit_transform(documents)

print(documents[0])
print("Bag-of-Words Matrix:")
for row in bow_matrix:
    print(row)

print("\nFeature Names:")
print(custom_bow.get_feature_names())
```
