```python
import torch
```

```python
t1 = torch.ones(2,3)
print("ndim:", t1.ndim)
print("shape:", t1.shape)
print(t1)

```

```python
t2 = torch.randn(2,3)
print("ndim:", t2.ndim)
print("shape:", t2.shape)
print(t2)
```

```python
torch.matmul(t1, t2.T)
```

```python
t2.T + 10
```

```python
tensor1 = torch.rand(2,2)
tensor2 = torch.rand(2,3)
print(f'{tensor1}\n{tensor2}')
print(f'{torch.matmul(tensor1, tensor2)}')
```
