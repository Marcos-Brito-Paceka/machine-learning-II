```python
# O que é um Tensor?

# Um tensor é a estrutura básica de dados usada no PyTorch 
# para representar números organizados em diferentes dimensões, 
# como escalares, vetores, matrizes ou conjuntos maiores de dados.
# Ele permite armazenar entradas, saídas, pesos e gradientes de 
# uma rede neural de forma eficiente, além de possibilitar cálculos
# rápidos em CPU ou GPU. No contexto de um MLP, tensores são
# essenciais porque os dados de entrada, os parâmetros das camadas 
# e os resultados intermediários precisam estar nesse formato para 
# que o modelo consiga fazer previsões, calcular o erro e ajustar 
# seus pesos durante o treinamento.

# Criando um tensor:
x = torch.tensor([1.0, 2.0, 3.0])
w = torch.tensor([0.5, 0.5, 0.5])

resultado = x * w
print(resultado)
```

```python
class MLPclassifier(torch.nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(MLPclassifier, self).__init__()
        self.camada1 = torch.nn.Linear(input_dim, hidden_dim)
        self.lrelu = torch.nn.LeakyReLU()
        self.camada2 = torch.nn.Linear(hidden_dim, output_dim)
    
    def forward(self, x):
        z1 = self.camada1(x)
        a1 = self.lrelu(z1)
        z2 = self.camada2(a1)
        return z2
    
    
```
